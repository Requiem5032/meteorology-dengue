```python
# Calculate mean best loss per country and save to average dir
import os
import yaml
import numpy as np

results_dir = '../results'
countries = [d for d in os.listdir(results_dir) if os.path.isdir(os.path.join(results_dir, d)) and not d.startswith('.')]

for country in countries:
    country_dir = os.path.join(results_dir, country)
    seeds = [d for d in os.listdir(country_dir) if d.startswith('seed_')]
    best_losses = []
    for seed in seeds:
        loss_result_path = os.path.join(country_dir, seed, 'loss_result.yaml')
        if os.path.exists(loss_result_path):
            with open(loss_result_path, 'r') as f:
                loss_data = yaml.safe_load(f)
                if isinstance(loss_data, dict) and 'best_loss' in loss_data:
                    best_losses.append(loss_data['best_loss'])
    if best_losses:
        mean_loss = float(np.mean(best_losses))
        avg_dir = os.path.join(country_dir, 'average')
        os.makedirs(avg_dir, exist_ok=True)
        with open(os.path.join(avg_dir, 'mean_best_loss.yaml'), 'w') as f:
            yaml.dump({'mean_best_loss': mean_loss}, f)
print('Mean best loss per country saved.')
```

```python
# Calculate average params per country and save to average dir
import os
import yaml
import numpy as np

results_dir = '../results'
countries = [d for d in os.listdir(results_dir) if os.path.isdir(os.path.join(results_dir, d)) and not d.startswith('.')]

for country in countries:
    country_dir = os.path.join(results_dir, country)
    seeds = [d for d in os.listdir(country_dir) if d.startswith('seed_')]
    param_list = []
    for seed in seeds:
        best_params_path = os.path.join(country_dir, seed, 'best_params.yaml')
        if os.path.exists(best_params_path):
            with open(best_params_path, 'r') as f:
                params = yaml.safe_load(f)
                # Exclude non-param keys
                param_dict = {k: v for k, v in params.items() if k not in ['best_loss']}
                param_list.append(param_dict)
    if param_list:
        # Find all param keys
        all_keys = set().union(*param_list)
        avg_params = {}
        for key in all_keys:
            values = [p[key] for p in param_list if key in p]
            if values:
                avg_params[key] = float(np.mean(values))
        avg_dir = os.path.join(country_dir, 'average')
        os.makedirs(avg_dir, exist_ok=True)
        with open(os.path.join(avg_dir, 'average_params.yaml'), 'w') as f:
            yaml.dump(avg_params, f)
print('Average params per country saved.')
```
